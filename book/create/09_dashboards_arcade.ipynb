## From Experience to Dashboard

Last week you assembled an experience from widgets. A dashboard uses the same architecture underneath: data sits on a server, and a browser client asks for it and draws it. What changes is the job the screen does. An experience invites someone to explore or follow a story. A dashboard puts several views of the same data on one screen so someone can monitor what is happening, make a decision, or spot a trend at a glance.

| What you did in Week 8 | The closest dashboard equivalent |
|---|---|
| Several widgets arranged on a page | Several elements arranged in one view |
| A Map widget connected to a data source | A map element drawn from a web map, whose layers can feed the other elements |
| Message actions between widgets | Actions between elements (a chart that moves the map, a map that filters a count) |
| Pages, and layout widgets that hold other widgets | Rows and columns that always fill the screen, plus an optional header and sidebar |

## What Is a Dashboard For?

A dashboard is a presentation of geographic information and data that helps you monitor events, make decisions, inform others, and see trends. Its elements work together on a single screen. Esri groups dashboards by purpose, and the right purpose follows from who is looking and what they need to do next.

| Type | Question it answers | Example |
|---|---|---|
| Operational | What is happening right now? | Current wildfire incidents on a map, with a count |
| Strategic | How are we doing against our goals? | A county office tracking service targets |
| Analytical | What trends and patterns are in the data? | Fire counts by year, compared with rainfall |
| Informative | What story does the data tell? | A public page explaining a fuels treatment project |

An effective dashboard does four things:

1. It draws attention to where it is needed.
2. It shows what matters most on a screen full of data.
3. It lets its audience understand what is happening and respond quickly.
4. It states measures clearly, accurately, and without distraction.

Dashboards also differ by how they are viewed. Some run unattended on a large screen in an operations center and offer a passive experience. Others are attended on a desktop monitor and invite more interaction.

## The Parts of a Dashboard

**Elements** are the building blocks. Common ones are maps, lists, charts, gauges, indicators, and tables. Most elements are data driven, which means each one needs a data source: a layer, a web map's layers, or a data expression (more on those below).

A practical habit: add a **map element first**. A web map's operational layers then show up as data sources when you configure the other elements.

Every element is configured through a few areas: a header for a title and more information, an optional top caption and bottom caption, the visualization itself, and an accessibility tab where you set the name a screen reader will read. Give each element a clear name early, because those names are what you pick from when you link elements together.

**Layout.** A view is built from rows and columns, and it always fills the screen. You can dock, resize, stack (tabs), and group elements. A view can also have one header and one sidebar.

**Views.** The desktop view is created first. A mobile view is a second layout for small screens, with its own arrangement of the same or different elements. Phones load the mobile view automatically when the screen is narrow, and a dashboard without one simply uses the desktop view. Phones have less room and memory, so a mobile view usually holds fewer elements.

## Actions: How Elements Talk to Each Other

One feature turns a set of charts on a page into a dashboard: **actions**. When actions are enabled, interacting with one element updates another. In Esri's wildfire tutorial, two links do most of the work:

- The **serial chart** is set to pan and zoom the map, so choosing a bar takes you to that fire.
- The **map** is set to filter the indicator, so the count shows only the fires inside the current map extent.

We practiced building these two links in class. If you have written an event listener in JavaScript, an action is the same idea without writing the code. It also matches the message actions you configured in Experience Builder last week.

## Design and Access

A **theme** is a set of colors and styling options saved with the dashboard and applied to both views. The light theme is the default, and you can choose a provided theme or customize one. A few basic colors drive most of the design, and the others are calculated from them. When the dashboard has a map, matching a light or dark basemap to the theme keeps the screen consistent.

Three habits keep a dashboard readable for the widest range of people:

- **Check contrast.** The color picker shows a warning when a text and background pairing fails a contrast check. Fix it instead of ignoring it.
- **Set accessible names.** Each element has an accessibility tab. The name you enter is what a screen reader announces, and it should be close to the element's title.
- **Use the glance test.** Look at your dashboard for five seconds, then look away. What do you remember? If it is the wrong thing, rearrange the screen so that the most important element is the most prominent one.

## Copying a Dashboard from Its Item ID

Every dashboard is an item with an ID. That makes it possible to start a new dashboard from an existing one, and it is a quick way to learn how a dashboard was built.

The pattern, as described in Esri's first tutorial:

1. Start creating a new dashboard (for example, from the **Create app** menu on a saved web map, then choose **Dashboards**).
2. In the address of the new dashboard, replace the text after `#id=` with the item ID of the dashboard you want to start from, and press Enter.
3. A **data source** window opens, with the original map, layers, and fields in one column and your replacements in the other.
4. Choose your own map and layers. Some fields have no one-to-one match in your data. For those, pick a different metric to show or delete the element.
5. If an element shows a ratio, set both a **value field** (the top of the fraction) and a **reference field** (the bottom).

This works when the creator has allowed it, so check the sharing settings, and a copy only shows data that you are permitted to see. A few questions to ask whenever you build on someone else's dashboard:

- Did I credit the person or organization whose work I started from?
- Am I allowed to reuse the data behind it?
- Which of their design choices did I inherit, and do they fit my audience?

Esri's community discussion ["How did they make that dashboard?"](https://community.esri.com/en/discussion/888765/how-did-they-make-that-dashboard) describes the technique.

## A First Look at Arcade

**Arcade** is a portable, lightweight, and secure expression language. It can do math, work with text, and evaluate logic. You may have seen it already in web maps, where it writes custom pop-ups, labels, and styles. Dashboards add two more uses.

| | Data expressions | Formatting expressions |
|---|---|---|
| What it does | Builds a new data source for the dashboard (combine, model, or enhance existing data) | Controls how an element looks, based on its data |
| What it returns | A **FeatureSet** | A **dictionary** of key-value pairs |
| Used by | Data-driven elements | Indicator, list, and table elements |
| Environment | The Dashboard data profile | A formatting profile for each element type |

Formatting is the easiest place to start. Say an indicator shows the number of current wildfires, and you want the number to turn red when the count is high. A formatting expression looks at the value, makes a decision, and returns the colors and text to display. A sketch of the shape:

```
// Arcade formatting expression for an indicator (illustrative sketch)
var count = $datapoint["fire_count"];
var color = "#2e7d32";

if (count >= 50) {
  color = "#b8312f";
} else if (count >= 20) {
  color = "#d98a00";
}

return {
  textColor: color,
  topText: "Current wildfires"
};
```

This is a sketch to show the shape of an expression. The names of the fields and the set of keys that an element accepts depend on the element, and Esri's formatting profile pages list them. You will work with real, tested expressions in class and in Lab 7.

## Arcade and the JavaScript You Know

As we have written JavaScript in this course, some of Arcade will feel familiar. Treat the match as a helpful similarity, with real differences.

| Looks familiar | Works differently |
|---|---|
| Variables, with `var` | Arcade runs inside ArcGIS applications, so there is no web page or DOM to change |
| `if` and `else` statements | There are no event listeners. The application decides when your expression runs |
| Functions and `return` | What you return has to match the structure the element expects (a dictionary, or a FeatureSet) |
| Comments with `//` | Many built-in functions are specific to GIS, such as ones that read feature attributes and filter layers |

When you meet a function you do not know, look it up in the Arcade function reference. Reading the reference is part of the skill, and it is how working GIS staff learn new functions too.

**Live in this notebook:** the cell below uses Python to show the same decision logic an Arcade formatting expression uses. The function reads a count, makes a decision, and returns a dictionary of display settings. The key names here are illustrative.

````python
from IPython.display import HTML, display

def indicator_format(fire_count):
    """Return a dictionary of display settings, the way an Arcade formatting expression does."""
    if fire_count >= 50:
        color, label = "#b8312f", "High activity"
    elif fire_count >= 20:
        color, label = "#d98a00", "Moderate activity"
    else:
        color, label = "#2e7d32", "Low activity"
    return {"textColor": color, "topText": "Current wildfires", "bottomText": label}

def show_indicator(fire_count):
    f = indicator_format(fire_count)
    display(HTML(
        f"<div style='font-family: Georgia, serif; border: 1px solid #999; padding: 10px; width: 220px; background: #f4f8fb;'>"
        f"<div style='font-size: 13px; color: #333;'>{f['topText']}</div>"
        f"<div style='font-size: 40px; font-weight: bold; color: {f['textColor']};'>{fire_count}</div>"
        f"<div style='font-size: 13px; color: #333;'>{f['bottomText']}</div></div>"
    ))

for count in (8, 27, 63):
    show_indicator(count)

# Try it: change the thresholds, or add a fourth level with its own color and label.
````
