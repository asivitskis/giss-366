## From Hand-Built to Builder-Built

For seven weeks you have built maps by writing or configuring each piece yourself. Experience Builder gives you prebuilt versions of many of those pieces and lets you assemble them by dragging and configuring. The architecture underneath has not changed: data sits on a server, and a browser client asks for it and draws it. What changes is how much of the client you author.

| What you did in this course | The closest Experience Builder equivalent |
|---|---|
| Your hosted feature layer and web map (Lab 1) | The data source a widget connects to |
| `L.map()` and a basemap tile layer | A Map widget, plus a Basemap Gallery widget if you want users to switch |
| Layer toggles and legends you hand-coded | Map Layers and Legend widgets |
| `index.html` structure and page layout | Pages, and layout widgets that hold other widgets |
| CSS rules for color, font, and spacing | A theme, with overrides on individual elements |
| JavaScript event listeners connecting one control to another | Message actions between widgets |
| Publishing to GitHub Pages | Publishing, plus choosing a sharing level |

The comparison is loose, and you should push on it in your Lab 6 reflection. Where does the analogy hold, and where does the builder decide something your code would have left to you?

## Plan Before You Build

Esri organizes the work into six steps: **Plan, Design, Configure, Customize, Deploy, Iterate**. The last three loop back on each other, so an app is revisited after launch and never finished once.

Planning starts with two questions: what problem is this app meant to solve, and who will use it? Everything else follows from those answers, including which widgets you need, which devices matter, and what data your audience is actually allowed to see. An app built without a clear purpose tends to collect features nobody asked for, and an app built without a clear audience tends to work for its builder only.

A useful planning habit is to write the answers down and keep them within reach while you build, then sketch a rough wireframe of the pages. The wireframe is a starting point that you expect to change.

*Esri reference: Course Doc 1, "Designing Web apps in ArcGIS Experience Builder," sections "ArcGIS Experience Builder workflow," "Ask questions," and "Mock up your design."*

## Design for People

Three design decisions shape how an app feels to use:

- **User experience and accessibility.** Can someone find what they need without instructions? Do images have text descriptions, can the app be used with a screen reader, and does the color scheme work for people with color vision differences?
- **Device format.** A phone user needs large tap targets and panels that collapse. A desktop user can handle always-visible menus and keyboard shortcuts. Decide which you are designing for, or plan for both.
- **Single page or scrolling.** Dashboards usually work best as one screen with details on demand. Stories, timelines, and FAQs usually work better as a scrolling read.

*Esri reference: Course Doc 1, "Functional web design" table in the Design section.*

## Builder Basics: Offerings, Modes, Templates, and Pages

You will use the **ArcGIS Online** offering in this course. Experience Builder also ships with **ArcGIS Enterprise** (which can include custom widgets made by your organization) and as a locally installed **developer edition** (which lets you build custom themes and widgets). The further you move toward the developer edition, the more control you gain and the more setup and skill you need.

When you create an experience you choose a **mode**. Full mode starts from a blank page or a template and supports multiple pages and windows. Express mode is a streamlined, single-page starting point. You can move from express to full mode, but not back, so choose with that in mind.

**Templates** are layouts with placeholder widgets for common cases such as dashboards, map tours, and timelines. A **page** holds your widgets, and an experience can have several. You will meet the interface panel by panel during the lab, so skim the Esri interface overview now and expect it to make more sense once you have the builder open.

*Esri reference: Course Doc 2, sections "Offerings," "Builder modes," "What are templates?," "What are pages?," and the interface overview ("Title bar" through "Right panel").*

## Widgets and Actions

**Widgets** are the building blocks of an experience. They fall into categories by purpose: map-centric, data-centric, page element, menu and toolbar, layout, and section. Layout and section widgets mostly hold and organize other widgets, which is part of how an experience adapts to different screens.

Each widget is set up through up to three parts:

- **Content:** what data the widget uses and what tools or arrangement it offers
- **Style:** how it looks
- **Action:** how it responds to users and to other widgets

Map-centric and data-centric widgets can use two kinds of action. A **message action** happens automatically when something triggers it. For example, searching for a place can move the map and shorten a list of nearby records in the same step. A **data action** adds a button that offers the user a menu of things they can do with the data. If you have written an event listener in JavaScript, a message action is the same idea without the code.

*Esri reference: Course Doc 3, "Using ArcGIS Experience Builder widgets," sections "Widget categories," "Widget configuration," and "Mapcentric and datacentric widget actions," and the widget list with practice scenarios on the second page.*

### Message action vs. data action, in miniature

*(Conceptual model only. This is not Esri code.)*

The cell below is a toy version of both action types. The message action fires automatically the moment the "search" happens, and it updates two other widgets in one step. The data action does nothing until the user picks something from the menu it offers.

```python
# A toy model of two widgets reacting to a third. Not Esri code, just the idea.

records = [
    {"name": "Silver City Library", "town": "Silver City"},
    {"name": "Gila Visitor Center", "town": "Silver City"},
    {"name": "Deming Water Plant",  "town": "Deming"},
    {"name": "Lordsburg Depot",     "town": "Lordsburg"},
]

map_widget  = {"center": None}
list_widget = {"shown": records}

# MESSAGE ACTION: runs automatically when the search widget publishes a message.
def on_search_message(place):
    map_widget["center"] = place
    list_widget["shown"] = [r for r in records if r["town"] == place]
    print(f"[message action] map -> {place}; list -> {[r['name'] for r in list_widget['shown']]}")

# DATA ACTION: offers a menu; nothing happens until the user chooses an item.
data_actions = {
    "Zoom to":  lambda rec: print(f"[data action] zoom map to {rec['name']}"),
    "Export":   lambda rec: print(f"[data action] export {rec['name']} as CSV"),
}

on_search_message("Silver City")                      # automatic: no click needed
print("Menu offered to user:", list(data_actions))     # waiting on the user...
data_actions["Zoom to"](list_widget["shown"][0])       # runs only because the user picked it

# Try it: search for "Deming" or "Lordsburg", or add a third data action such as "Share".
```

## Themes and Customizing

A **theme** is a shared set of design choices (colors, fonts, button styles) applied across the experience. You can override the theme on any page, widget, or element. If that sounds like CSS, it should: a general rule sets the default, and a more specific rule wins. Themes also make it easier to keep several experiences looking alike inside one organization.

Customizing covers two things: functionality (adding actions that make the app easier to use) and visuals (images, theme, typeface). Configuring gets the features working. Customizing makes them usable.

**Live in this notebook:** the cell below uses the CSS you learned in Week 4 to mimic a theme with one override. The general rule styles every button, and the more specific rule wins for the one button that carries the extra class.

*Esri reference: Course Doc 4, "Customizing and Deploying web apps," page on themes, and Course Doc 1, section "Customize."*

```python
from IPython.display import HTML, display

theme_demo = """
<style>
  .eb-theme { font-family: Georgia, serif; color: #1a3c5e; border: 1px solid #999;
              padding: 10px; background: #f4f8fb; }
  /* The "theme": a general rule that styles every button */
  .eb-theme .eb-btn { background: #0079c1; color: white; border: none; border-radius: 3px;
                      padding: 6px 12px; margin: 4px; font-family: inherit; }
  /* The "override": a more specific rule that wins for one element */
  .eb-theme .eb-btn.eb-override { background: #b8312f; font-weight: bold; }
</style>
<div class="eb-theme">
  <p>Widgets inherit the theme's font and color.</p>
  <button class="eb-btn">Search</button>
  <button class="eb-btn">Legend</button>
  <button class="eb-btn eb-override">Report a problem (overridden)</button>
</div>
"""

display(HTML(theme_demo))

# Try it: change the theme's background color, or add a second .eb-override button.
```

## Deploying and Iterating

Esri's deployment sequence is **preview, update sharing, deploy, test, refine**. Two points matter most.

First, **sharing has a floor.** An experience and every web map and data source it uses must be shared at the same level or wider. The four levels are public (anyone with the link), organization (members of your ArcGIS Online organization), group (members of a chosen group), and owner (only you). Publishing an experience at one level while its data sits at a narrower level is a common reason an app opens for its builder and fails for everyone else.

Second, **test like a visitor.** After publishing, open the app with a different account and on a different device. Preview inside the builder cannot show you what a stranger sees.

Iteration continues after launch: gather feedback, fix problems, and republish.

**Live in this notebook:** the cell below checks the sharing floor for a few made-up experiences. The builder owner can always open everything, which is exactly why this problem hides from you.

*Esri reference: Course Doc 4, "Customizing and Deploying web apps," the deployment page ("Preview," "Update sharing," "Deploy," "Test," "Refine") and the iteration page.*

```python
# Sharing levels, from narrowest to widest.
LEVELS = ["owner", "group", "organization", "public"]
rank = {lvl: i for i, lvl in enumerate(LEVELS)}

def check_sharing(experience, **items):
    """Every item the experience uses must be shared at the experience's level or wider."""
    print(f"Experience shared at: {experience}")
    problems = []
    for name, level in items.items():
        ok = rank[level] >= rank[experience]
        print(f"  {name:<12} {level:<13} {'ok' if ok else 'TOO NARROW'}")
        if not ok:
            problems.append(name)
    if problems:
        print(f"  -> Opens for the builder, but visitors will see a blank or broken app ({', '.join(problems)}).\n")
    else:
        print("  -> Visitors at this level can load everything.\n")

check_sharing("organization", web_map="organization", feature_layer="organization")
check_sharing("public",       web_map="public",       feature_layer="owner")
check_sharing("organization", web_map="group",        feature_layer="organization")

# Try it: change the levels, or write the check for the case you would use for a class project
# that you want your instructor to open.
```

## The Continuum: How Much Do You Want to Build?

Experience Builder is one point on a longer line. This course has covered both ends of it.

| Tier | You control | You need | Who can realistically build and maintain it |
|---|---|---|---|
| No-code builder (Experience Builder in ArcGIS Online) | Layout, widgets, theme | An organizational account | Anyone comfortable with the interface |
| Low-code customization of ArcGIS Online items | Expressions inside pop-ups, labels, and symbology | The same account, plus light scripting | People who can read and write short scripts |
| Developer edition of Experience Builder | Custom widgets and themes | A local install, JavaScript and React skills, a place to host | Developers |
| ArcGIS Maps SDK for JavaScript | Nearly everything in the client | Programming skills, hosting, licensing | Developers |
| Open-source libraries (Leaflet, MapLibre) | Everything in the client | Programming skills and hosting | Developers, with no vendor license |

This week only the first row is hands-on. The others are named so you can see where each choice leads. Moving down the table buys control and costs skill, setup, and time. Moving up the table buys speed and polish and costs flexibility. Neither direction is correct in general, and the right choice depends on the audience and the team.
