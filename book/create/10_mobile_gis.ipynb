## What Is Mobile GIS?

Mobile GIS brings GIS to the place being mapped. A device that knows where it is can record what you see where you see it. A collector is often standing in the subject area, so they can see, touch, hear, smell, and feel what they are recording.

### Advantages

- **Mobility:** the tool goes where the subject is
- **Location awareness:** positioning technologies give a device its location. These include GPS, cellular networks, Wi-Fi networks, and Bluetooth
- **The user is in the subject area:** the person recording has direct sensory knowledge of what they are mapping
- **Near real-time information:** a record can reach a map or dashboard within moments
- **A large volume of users:** many people can contribute at once
- **A versatile means of communication:** a phone carries text, photos, and location in one submission

### Supporting technologies

Mobile GIS depends on four kinds of technology working together: mobile devices, mobile operating systems, wireless communication technologies, and positioning technologies. A weak link in any of them shows up in the field as a missed GPS fix, a dropped signal, or a map that will not load.

## Volunteered Geographic Information and iNaturalist

After you submitted your observation, you compared it with the ten most recent iNaturalist observations near you. iNaturalist is a form of **citizen science**, and the data it produces is **volunteered geographic information (VGI)**.

> **VGI** is digital spatial data produced voluntarily by citizens rather than by formal institutional data producers.

VGI is a form of crowdsourcing that engages people to share observations on anything. Common examples include natural observations and invasive species records (iNaturalist), and accidents and police locations reported by mobile users (Waze). A growing number of recent scientific publications cite iNaturalist data, including local studies in entomology.

- [iNaturalist observations map](https://www.inaturalist.org/observations?subview=map)

### Tools for collecting VGI

Survey123 works well for VGI because it does not require named users, so a person can contribute without being added to your organization. ArcGIS QuickCapture is another good Esri option. Citizen science platforms and open source providers fill the same role, and the open source side is covered later on this page.

### Who contributes?

VGI comes from people who choose to contribute. That choice shapes the data.

| | VGI (iNaturalist, a public Survey123 form) | Organizational field data (a campus inventory) |
|---|---|---|
| **Who contributes** | Volunteers, who often self-select | Trained staff or a defined team |
| **Who designs the form** | The platform or the project lead | The team doing the work |
| **Strengths** | Many contributors, wide coverage, fast arrival of records | Consistent fields, controlled choice lists, known collectors |
| **Limits** | Uneven coverage, and gaps where few people live, visit, or have device access and cell coverage | Smaller coverage, and a dependence on the capacity of one team |

**Think about it:** who is likely to contribute in your area, and who might be missing? What could you safely conclude from a map of those records, and what could you not?

## Feature Layers: The Backbone

Feature layers sit behind Esri mobile GIS products. They are the same kind of hosted layer you built in Lab 1 and used as a data source in your Week 9 dashboards. One layer can serve many products at once:

| Product | What it does with the layer |
|---|---|
| **Survey123 form** | Adds new records through a smart form |
| **Field Maps** | Adds and edits points, lines, and areas using the map or GPS, with a map-driven form |
| **Web map** | Displays the records, with pop-ups that show the form fields and photos |
| **Dashboard** | Summarizes the records with indicators, lists, and charts |

### Views

A **view** and its **source feature layer** reference the same data. Each view can independently set:

- how it is **shared** with others
- how it is **drawn** (symbology)
- which features are **displayed** (filtering)
- whether the layer can be **edited**

This is how one dataset can be editable for the collection team and read-only for the public, with sensitive features filtered out of the public view.

## Who Designs the Form?

A form decides, before anyone collects a single point, what a collector is allowed to record. The two forms below ask about the same park bench. They are plain HTML, so you can click through them here.


```python
from IPython.display import HTML, display

form_demo = """
<style>
  .fd { font-family: Georgia, serif; color: #1a3c5e; display: flex; gap: 16px; flex-wrap: wrap; }
  .fd .card { flex: 1; min-width: 260px; border: 1px solid #999; padding: 12px; background: #f4f8fb; }
  .fd h4 { margin: 0 0 6px 0; }
  .fd label { display: block; margin-top: 10px; font-weight: bold; }
  .fd input, .fd select { width: 100%; padding: 5px; margin-top: 3px; font-family: inherit; box-sizing: border-box; }
  .fd .req { color: #b8312f; }
  .fd small { display: block; margin-top: 8px; color: #555; }
</style>
<div class="fd">
  <div class="card">
    <h4>Form A: free text</h4>
    <label>What is it?</label>
    <input type="text" placeholder="Type anything">
    <label>What shape is it in?</label>
    <input type="text" placeholder="Type anything">
    <small>Collectors could enter: bench, Bench, benches, picnic seat, wood thing by the path. Every answer is possible, and every answer needs cleaning before you can count anything.</small>
  </div>
  <div class="card">
    <h4>Form B: choice lists and a required field</h4>
    <label>Amenity type <span class="req">* required</span></label>
    <select>
      <option>Choose one</option><option>Bench</option><option>Picnic table</option>
      <option>Trash can</option><option>Restroom</option><option>Other</option>
    </select>
    <label>Condition <span class="req">* required</span></label>
    <select>
      <option>Choose one</option><option>1 (unusable)</option><option>2 (poor)</option>
      <option>3 (good)</option><option>4 (excellent)</option>
    </select>
    <label>Accessibility note</label>
    <input type="text" placeholder="Optional, with a hint for the collector">
    <small>Every record can be counted and compared. A collector who finds a drinking fountain has to choose Other, or leave it out.</small>
  </div>
</div>
"""

display(HTML(form_demo))

# Try it: add an option such as "Drinking fountain" to Form B's amenity list,
# or change the condition scale to three levels, and rerun the cell.
```

Form A keeps everything a collector wants to say. Form B keeps the data consistent. Neither is correct in general: a choice list makes records comparable, and it also decides which kinds of things exist in your dataset.

## Three Ways to Build a Mobile App

Browser-based, native, and hybrid apps all put maps on phones. Each carries benefits and tradeoffs.

| Approach | How it works | Examples | Typical benefit | Typical tradeoff |
|---|---|---|---|---|
| **Browser-based** | Runs on the web inside the phone's browser | ArcGIS Web Apps | Nothing to install, and one version for everyone | Limited access to device features, and a weaker offline story |
| **Native** | Installed on the device | ArcGIS Field Maps, Survey123, QField | Full use of GPS, camera, and offline maps | Must be installed, and kept updated per platform |
| **Hybrid** | Native components combined with HTML, JavaScript, and CSS | The simplest version loads HTML and JavaScript content inside a native app. More advanced versions use frameworks for deeper integration with the native platform | Reuses web skills and code | Sits between the other two, and depth of integration varies |

## Esri Apps for Mobile GIS

Esri offers several apps for different kinds of field work.

| | **Survey123** | **ArcGIS QuickCapture** | **ArcGIS Field Maps** |
|---|---|---|---|

## Open Source Option: QField

The Esri apps are one family of tools. There is an open source family too, and **QField** is the one to know for field data collection.

QField is an open source app for field work that connects to your **QGIS** projects. It works fully offline or connected, and it is released under the GPL, which means it is free to use and modify. It is available for Android and iOS, and it can also be used on the desktop (Windows, macOS, and Linux).

### How the workflow runs

1. **Design in QGIS on a computer.** You set up your layers, forms, and symbology in a QGIS project.
2. **Package the project with the QFieldSync plugin.** The plugin checks your project and suggests changes so that it works well in the field.
3. **Move the project to a device.** One route is **QFieldCloud**, an optional cloud service that synchronizes data between your phone and your desktop. The QFieldCloud backend is itself open source. You can also transfer a project to the device without the cloud.
4. **Collect data in QField.** The interface is simplified and touch-friendly, and it supports GPS positioning and camera integration.
5. **Synchronize back** to the project on your computer.

### QField and Field Maps side by side

| | **QField** | **ArcGIS Field Maps** |
|---|---|---|
| **Authoring** | QGIS on a computer, plus the QFieldSync plugin | Field Maps Designer in ArcGIS Online |
| **Where the data lives** | In your project, which can link to external databases or use simple geospatial formats, or in QFieldCloud | In a hosted feature layer in ArcGIS Online |
| **Cost and license** | Open source and free to use | Needs an ArcGIS account with the right role |
| **Path to the web** | You choose and build the web products yourself | The same feature layer feeds a web map, an experience, or a dashboard directly |
| **What you need to know** | QGIS skills | ArcGIS Online skills |
| **Who controls the stack** | You do, from the data format to the code | Esri's platform sets the boundaries |

### Why this connects to the rest of the course

Early in the course you built maps with Leaflet and MapLibre, which are open source libraries. QField is the field collection counterpart. The pattern from Experience Builder week holds here as well: moving toward open source buys control and costs setup and skill, and moving toward the Esri products buys speed and integration and costs flexibility.

Reasons a project might choose QField include a team that already works in QGIS, a limited budget for software licenses, and a need to keep data on servers the team controls. Reasons to choose Field Maps include an organization that already runs on ArcGIS Online, and a goal of getting from field to dashboard quickly.

Other open source and community tools exist for field data collection, including KoboToolbox, Open Data Kit, and Mergin Maps. They are worth knowing by name if your project needs something QField does not offer.

**Optional:** to explore QField, start with the Get Started pages in the [QField documentation](https://docs.qfield.org). Lab 8 itself uses Field Maps.

## Case Study: Mobile GIS at WNMU

How can mobile GIS support a real project at our university? WNMU is taking an inventory of its local plant life using mobile GIS approaches. University staff have used two different tools for it.

### TreePlotter and Field Maps

**TreePlotter** is part of a larger statewide initiative led by the state's urban forestry department to help local communities create urban tree inventories. It is an independent third-party application, and the team used it because it lets many communities gather and share data in one place. The statewide baseline from the New Mexico Urban Tree Inventory (2017 to 2018) is shared through TreePlotter.

**ArcGIS Field Maps** was chosen by the Arboretum team for the campus shrub survey, because it gives more granular control over what data is collected and how it is stored and accessed.

| | **TreePlotter** | **ArcGIS Field Maps** |
|---|---|---|
| **Used at WNMU for** | Campus trees, within the statewide inventory | The campus shrub survey |
| **What it favors** | Coordination: many communities, one shared dataset | Control: the team decides what is collected, where it is stored, and who can see it |
| **Who designs the form** | The statewide initiative and the application | The Arboretum team |
| **Cost of the choice** | The team accepts fields it did not choose | The team takes on the design and upkeep of its own form and data |

Choosing a platform is a decision about who collects data, who can use it, and what questions it can answer.
