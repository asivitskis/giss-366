## What Is Spatiotemporal Data?

**Spatiotemporal data** answers three questions at once:

- **Where** is it? (a location)
- **When** was it there? (a timestamp)
- **What** is it, or what was it doing? (attributes)

A single GPS fix answers all three: *this eagle, at this place, at this time.* Many fixes in a row make a track, and a track lets you ask about movement, such as when an animal left, how far it went, and how fast it traveled.

### Archived, Temporal, and Live

| Type | What it is | Example |
|---|---|---|
| **Static** | A snapshot with no time dimension that matters | A layer of campus buildings |
| **Temporal** | Data that records when things happened. It can be archived and still be temporal | The golden eagle tracks from 2002 to 2003 |
| **Live** | Temporal data that keeps arriving, so the app has to refresh on a timer or a connection | A vehicle feed or a real-time sensor |

The eagle data you use this week is **archived** and **temporal**. It is not live. Nothing in the page updates itself. A live version would add a refresh step, with a page that fetches new data on a timer, and we mention it as a direction without building it.

### Anatomy of a Tracking Record

A tracking record is small. The minimum is an animal ID, a timestamp, a longitude, and a latitude. Everything else is an optional attribute. The same record can be stored as a row in a CSV file or as a feature in a GeoJSON file.

The values below are **illustrative**, made up to show the structure. They are not rows from the eagle layer.

| animal_id | timestamp | longitude | latitude | daily_km |
|---|---|---|---|---|
| 37300a | 2002-10-14 06:00 | -105.20 | 40.10 | |
| 37300a | 2002-10-16 18:00 | -105.31 | 39.72 | 18.4 |
| 37300a | 2002-10-19 06:00 | -105.62 | 39.15 | 26.7 |
| 37301a | 2002-10-15 12:00 | -109.40 | 44.80 | |
| 37301a | 2002-10-18 00:00 | -109.52 | 44.31 | 15.9 |

The same kind of record as a GeoJSON feature:

```json
{
  "type": "Feature",
  "geometry": { "type": "Point", "coordinates": [-105.31, 39.72] },
  "properties": {
    "animal_id": "37300a",
    "timestamp": "2002-10-16T18:00:00Z",
    "daily_km": 18.4
  }
}
```

Notice that `daily_km` is empty for each animal's first fix. It is calculated from the interval between two fixes, so the first fix has nothing to compare with. A missing value in a chart-ready field is a design fact to plan for, not an error.

### The Recording Schedule Shapes the Data

A GPS tag does not watch the animal continuously. It records a location on a schedule, and the schedule is a choice. The golden eagle layer holds **359 Argos satellite fixes from four eagles**, arriving about every 2 to 3 days between October 2002 and September 2003. That has consequences:

- A count of points reflects the recording schedule as well as the animal's behavior. More points in a month does not mean the eagle spent more time there.
- A short time window may hold only a few points per eagle, so an average calculated from them can swing widely.
- Thinning the data (keeping fewer fixes) was a decision someone made. A different decision would change every count in the app.

## The Big Idea: A Slider Is a Filter

> **A time slider is a filter on a timestamp field.** Moving the slider changes a cutoff or a window, and everything connected to the data redraws to show only the records that pass.

When time does not work, one of three things is wrong. Ask these in this order, in any tool:

1. **Is the timestamp a real date?**
2. **Has the app been told to treat it as time?**
3. **Which elements are listening to it?**

### The Crosswalk: One Idea, Two Apps

The same three questions apply to ArcGIS and to code. Keep this table open when you read the code later on this page.

| **Question** | **ArcGIS Online / Experience Builder** | **MapLibre and Chart.js** |
|---|---|---|
| **1. Is the timestamp a real date?** | The field is a date type in the layer, not text | `Date.parse` in step 1 turns the timestamp into a number. A bad value gives `NaN` |
| **2. Has the app been told to treat it as time?** | Time is enabled on the layer, and the Timeline widget is connected to the Map | The slider's `min`, `max`, and `step` are set from the first and last time in step 2 |
| **3. Which elements are listening?** | The map, the Text widget, the Filter, and the Chart, if each is connected to the same time-enabled data source | Everything that draws from `shown` in `update()`: the map, the count, and the chart |
| **Time slider** | Timeline widget | Date slider (`<input type="range">`) |
| **Indicator** | Text widget with dynamic content (a count) | `$("count").textContent = shown.length` |
| **Chart** | Chart widget | Chart.js `drawChart` function |

### Fields to Consider for Your Chart

The layer has about 30 fields. These seven are the most useful for a chart that updates as the timeline moves.

| Field | What it records | Chart idea |
|---|---|---|
| `animal_id` | ID of each tagged eagle (37300a, 37301a, 37303a, 37433a) | One bar, line, or color per eagle |
| `latitude` | North-south position of each fix, about 30° to 53° N across the layer | Mean latitude per eagle within the time window |
| `month_label` | Month of the fix as text with the number first (for example, "10 Oct"), so months sort in calendar order | Count of fixes per month, split by `animal_id` |
| `season` | Fall, Winter, Spring, or Summer, assigned by calendar month | Mean `daily_km` by season |
| `daily_km` | Estimated daily travel rate for the interval between two fixes. Empty for each eagle's first fix | Average or maximum per eagle |
| `movement_class` | Start, Resident (under 10 km/day), Local (10 to 50), Moving (50 to 150), or Long distance (150 and up) | Count of fixes by movement class in the window |
| `dist_from_start_km` | Straight-line distance from the eagle's first recorded location | Maximum or mean per eagle |

Three sample questions a chart could answer:

- Which eagle sits farthest north or south during the window you selected? (mean `latitude` by `animal_id`)
- Do eagles travel more per day in some seasons than others? (mean `daily_km` by `season`)
- When does each bird leave its starting area? (maximum `dist_from_start_km` by `animal_id`)

### Interpret with Care

- Fixes arrive about every 2 to 3 days, so a short time window may hold only a few points per eagle. Averages from a few points can swing widely.
- Eagle 37301a stops transmitting in early February 2003, so it has no spring or summer data. A viewer could mistake a silent tag for an eagle that stopped moving, or for one that died. The data does not say which.
- Argos locations carry error. About a third of fixes are accurate only to within 1 km or worse, or have no error estimate (see `location_accuracy`), and that error can inflate distance and speed values.
- Many charts sort categories alphabetically by default. Check the order of your categories, especially for `movement_class`.

### Reading the Page by Its Five Steps

Everything on the page is drawn from **one array called `shown`**: the locations up to the slider date. That is why the map, the count, and the chart always agree. The script is numbered 1 to 5.

| Step | What it does | Which question it answers |
|---|---|---|
| **1. Load the data** | Reads `tracks.geojson` and turns each timestamp into a number with `Date.parse` | Question 1: is the timestamp a real date? |
| **2. Set up the page text and slider** | Fills in the title and question, and sets the slider's `min`, `max`, and `step` from the first and last time | Question 2: has the app been told to treat it as time? |
| **3. Build the map** | Creates the MapLibre map and its point layer | |
| **4. Draw the chart** | The `drawChart` function builds a Chart.js chart. This is where you choose your visualization | |
| **5. `update()`** | Runs every time the slider moves. Keeps the points where `p.time <= cutoff`, then sends them to the map, the count, and the chart | Question 3: which elements are listening? |

A simplified sketch of the pattern in step 5 (the real template's function names differ):

```javascript
function update() {
  const cutoff = Number(slider.value);                 // where the slider is now
  const shown = points.filter(p => p.time <= cutoff);  // the one filter
  drawMap(shown);                                      // map listens
  $("count").textContent = shown.length;               // count listens
  drawChart(shown);                                    // chart listens
}
```

### How to Use the Code Cells

Each example below is a Python cell holding a block of JavaScript. **Edit the JavaScript, then re-run the cell** to see the chart change. The cells run in your browser, so you need an internet connection to load the Chart.js library. If something breaks, re-run the cell you started from.

Run this cell first. It sets up a helper that draws a chart in a small frame.

````python
from IPython.display import HTML, display
import html as _html

# Chart.js is loaded from a CDN in each frame. The version is pinned so the
# examples keep working. Change the number to test another version.
CHARTJS = "https://cdn.jsdelivr.net/npm/chart.js@4.4.1/dist/chart.umd.min.js"

def chart_demo(js, body="", height=400):
    """Draw a Chart.js example in its own frame.
    js   : the JavaScript to run. The canvas has id="myChart".
    body : optional HTML placed above the canvas (for example, a slider)."""
    page = f"""<!DOCTYPE html><html><head><meta charset="utf-8">
<style>body{{font-family:Georgia,serif;margin:10px;color:#1a3c5e}}
#wrap{{max-width:640px}} input[type=range]{{width:100%;margin-bottom:8px}}</style></head>
<body><div id="wrap">{body}<canvas id="myChart"></canvas></div>
<script src="{CHARTJS}"></script>
<script>
{js}
</script></body></html>"""
    src = _html.escape(page, quote=True)
    display(HTML(f'<iframe srcdoc="{src}" width="100%" height="{height}" '
                 f'style="border:1px solid #999;"></iframe>'))
````

### The Big Idea: One Config Object

A Chart.js chart is **one config object with three parts**:

| Part | What it controls | Example |
|---|---|---|
| **`type`** | The kind of chart | `"line"`, `"bar"`, `"pie"`, `"doughnut"`, `"scatter"` |
| **`data`** | What gets drawn: the labels and one or more datasets | `labels`, `datasets`, each dataset's `data`, `label`, and colors |
| **`options`** | How it looks and behaves | Title, legend, axis limits, tooltips |

The call has the same shape every time: `new Chart(canvas, { type, data, options })`. The example below is a line chart of house prices against house size, the same one used in the W3Schools introduction. The comments mark the three parts.

> **Version note:** Chart.js changed how options are written between version 2 and version 3. Older tutorials, including some W3Schools examples, may put `legend` and `title` directly inside `options` and use `yAxes`. Current versions put them under `options.plugins` and use `scales.y`. This page uses a current version. If an example from another site does not behave, check the version in its `<script>` tag first.

````python
step0 = r"""
const sizes  = [50, 60, 70, 80, 90, 100, 110, 120, 130, 140, 150];
const prices = [7, 8, 8, 9, 9, 9, 10, 11, 14, 14, 15];

new Chart(document.getElementById("myChart"), {
  type: "line",                          // PART 1: type
  data: {                                // PART 2: data
    labels: sizes,
    datasets: [{ data: prices, borderColor: "blue", backgroundColor: "blue" }]
  },
  options: {                             // PART 3: options
    plugins: {
      legend: { display: false },
      title: { display: true, text: "House Prices vs. Size", font: { size: 16 } }
    }
  }
});
"""

chart_demo(step0)
````

#### Warm-Up Step 3: Use an Array of `{ x, y }` Objects

So far the data was typed in as two lists. Real data usually arrives as **a list of records**, one object per row, like the rows of a tracking file. The code below starts from that shape and turns it into chart data.

It also changes the chart `type` to `"scatter"` with `showLine: true`. A line chart places its labels at equal spacing, while a scatter chart places each point at its true `x` value. This matters for time: if fixes arrive every 2 days and then every 10 days, a time chart should show the gap.

````python
step3 = r"""
const groups  = ["Suburb", "City"];
const colorOf = { Suburb: "#1b9e77", City: "#d95f02" };

const points = [
  { group: "Suburb", x: 50,  y: 7 },  { group: "Suburb", x: 60,  y: 8 },
  { group: "Suburb", x: 100, y: 9 },  { group: "Suburb", x: 150, y: 15 },
  { group: "City",   x: 50,  y: 9 },  { group: "City",   x: 60,  y: 10 },
  { group: "City",   x: 100, y: 12 }, { group: "City",   x: 150, y: 20 }
];

const ctx = document.getElementById("myChart");

new Chart(ctx, {
  type: "scatter",
  data: {
    datasets: groups.map(g => ({
      label: g,
      data: points.filter(p => p.group === g).map(p => ({ x: p.x, y: p.y })),
      borderColor: colorOf[g],
      backgroundColor: colorOf[g],
      showLine: true
    }))
  },
  options: {
    plugins: {
      title: { display: true, text: "House Prices vs. Size", font: { size: 16 } }
    },
    scales: { x: { min: 50, max: 150 } }
  }
});
"""

chart_demo(step3)
````

**You should see:** two lines with points at their true sizes (50, 60, 100, 150), so the gaps are uneven.

**Answer for yourself:** find the line that turns the list of objects into chart data. What does `.filter` do, and what does `.map` do?

> **The two methods to know.** `.filter(test)` keeps only the items that pass a test and returns a shorter list. `.map(change)` returns a new list in which every item has been changed. Here, `.filter` picks out one group's rows, and `.map` reshapes each row into the `{ x, y }` object Chart.js wants. In the app, the same `.filter` call is how a slider works.

#### Warm-Up Step 4: Filter Before Drawing

Now add a **cutoff**. Two new lines make a list called `shown`, holding only the points at or below the cutoff, and the chart draws from `shown` instead of `points`. This is the idea from the whole week in two lines.

````python
step4 = r"""
const groups  = ["Suburb", "City"];
const colorOf = { Suburb: "#1b9e77", City: "#d95f02" };

const points = [
  { group: "Suburb", x: 50,  y: 7 },  { group: "Suburb", x: 60,  y: 8 },
  { group: "Suburb", x: 100, y: 9 },  { group: "Suburb", x: 150, y: 15 },
  { group: "City",   x: 50,  y: 9 },  { group: "City",   x: 60,  y: 10 },
  { group: "City",   x: 100, y: 12 }, { group: "City",   x: 150, y: 20 }
];

const ctx = document.getElementById("myChart");

const cutoff = 100;                                     // try 60, 100, 150
const shown  = points.filter(p => p.x <= cutoff);      // the filter

new Chart(ctx, {
  type: "scatter",
  data: {
    datasets: groups.map(g => ({
      label: g,
      data: shown.filter(p => p.group === g).map(p => ({ x: p.x, y: p.y })),
      borderColor: colorOf[g],
      backgroundColor: colorOf[g],
      showLine: true
    }))
  },
  options: {
    plugins: {
      title: { display: true, text: "House Prices vs. Size", font: { size: 16 } }
    },
    scales: { x: { min: 50, max: 150 } }
  }
});
"""

chart_demo(step4)
````

Try cutoff values of 60, 100, and 150. In the eagle app, `x` is a time and the cutoff is the slider's date.

**You should see:** only the points up to the cutoff.

**Answer for yourself:** what is `shown`, and what draws from it?

#### Warm-Up Step 5: Put the Chart in a Function

A slider calls for a chart to be drawn **many times**, once per move. A function lets you do that. The new part is `chart.destroy()`, which removes the old chart before a new one is built.

````python
step5 = r"""
const groups  = ["Suburb", "City"];
const colorOf = { Suburb: "#1b9e77", City: "#d95f02" };

const points = [
  { group: "Suburb", x: 50,  y: 7 },  { group: "Suburb", x: 60,  y: 8 },
  { group: "Suburb", x: 100, y: 9 },  { group: "Suburb", x: 150, y: 15 },
  { group: "City",   x: 50,  y: 9 },  { group: "City",   x: 60,  y: 10 },
  { group: "City",   x: 100, y: 12 }, { group: "City",   x: 150, y: 20 }
];

const ctx = document.getElementById("myChart");

let chart;                                   // remembers the current chart

function drawChart(shown) {
  if (chart) chart.destroy();                // remove the old chart first
  chart = new Chart(ctx, {
    type: "scatter",
    data: {
      datasets: groups.map(g => ({
        label: g,
        data: shown.filter(p => p.group === g).map(p => ({ x: p.x, y: p.y })),
        borderColor: colorOf[g],
        backgroundColor: colorOf[g],
        showLine: true
      }))
    },
    options: {
      plugins: {
        title: { display: true, text: "House Prices vs. Size", font: { size: 16 } }
      },
      scales: { x: { min: 50, max: 150 } }
    }
  });
}

const cutoff = 100;
const shown  = points.filter(p => p.x <= cutoff);
drawChart(shown);
"""

chart_demo(step5)
````

**You should see:** the same chart as Step 4.

**Answer for yourself:** why does the function start with `chart.destroy()`? (Hint: think about what happens each time a slider moves.)


#### ChartJS Examples

Changing the `type` changes what a viewer can see. The lab offers three chart styles for the eagle data. Each answers a different kind of question.

| Chart | Type in Chart.js | A question it can answer | What it cannot show |
|---|---|---|---|
| **Line** | `"line"` | How does a value change over the time window? What is the average range of movement? | Counts of categories, or where an animal is on the map |
| **Pie (counts)** | `"pie"` | For which bird do we have the most recorded observations? | Change over time, or which fixes are closest to each other |
| **Gauge (a modified doughnut)** | `"doughnut"` | Where is an animal at ___% through the observation window? | Anything beyond one number against a total |

The three cells below use **made-up numbers** so you can see the shapes. They are not the values in the eagle layer. Change `type` in the middle cell from `"pie"` to `"doughnut"` and see what happens: a pie becomes a ring with the same data.

````python
line_js = r"""
new Chart(document.getElementById("myChart"), {
  type: "line",
  data: {
    labels: ["Oct", "Nov", "Dec", "Jan", "Feb", "Mar"],
    datasets: [{
      label: "Mean daily km (made-up numbers)",
      data: [22, 30, 12, 9, 10, 41],
      borderColor: "#1b9e77", backgroundColor: "#1b9e77"
    }]
  },
  options: { plugins: { title: { display: true, text: "Line: change over time" } } }
});
"""
chart_demo(line_js, height=370)
````

````python
pie_js = r"""
new Chart(document.getElementById("myChart"), {
  type: "pie",                  // try "doughnut"
  data: {
    labels: ["37300a", "37301a", "37303a", "37433a"],
    datasets: [{
      data: [110, 52, 98, 99],
      backgroundColor: ["#1b9e77", "#d95f02", "#7570b3", "#e6ab02"]
    }]
  },
  options: { plugins: { title: { display: true, text: "Pie: share of fixes (made-up numbers)" } } }
});
"""
chart_demo(pie_js, height=440)
````

````python
gauge_js = r"""
const done = 62;                // percent through the observation window

new Chart(document.getElementById("myChart"), {
  type: "doughnut",
  data: {
    labels: ["Elapsed", "Remaining"],
    datasets: [{
      data: [done, 100 - done],
      backgroundColor: ["#1b9e77", "#dddddd"]
    }]
  },
  options: {
    rotation: -90,              // start on the left
    circumference: 180,         // draw only the top half
    plugins: { title: { display: true, text: "Gauge: " + done + "% through the window" } }
  }
});
"""
chart_demo(gauge_js, height=330)
````

A gauge is a doughnut with two settings changed. It shows how far an eagle has come in its observation window, one number against a total. That is a good choice for a single indicator and a poor one for comparing four birds.

**Think about it:** choosing a chart is choosing what a viewer will notice. Which of the three would you add beside the eagle map to help an audience answer *your* question? What might a viewer wrongly conclude from it?

#### Common Chart.js Problems

| What you see | Likely cause |
|---|---|
| Nothing appears, and the Console says `Chart is not defined` | The Chart.js `<script>` tag is missing, misspelled, or loads after your code |
| The console says the canvas is already in use | A new chart was built on the same canvas without `chart.destroy()` first |
| Your title or legend options are ignored | The options are written for another version. Current versions use `options.plugins.title` and `options.plugins.legend` |
| The x-axis spacing looks even, but the times are not | A `"line"` chart spaces its labels equally. Use `"scatter"` with numeric `x` values (or a time scale) to show true gaps |
| Numbers sort or add as text ("10" before "9") | The values are strings. Convert them with `Number(...)` |
| Categories appear in the wrong order | The chart uses the order of your labels. Sort the labels yourself, or put a number first, as `month_label` does ("10 Oct") |
| The chart is a blank box | The canvas has no room. Put it in a container with a width, or check that the data list is not empty |
| Data and labels do not line up | The lists have different lengths, or `labels` and `data` came from different filters |

## Who Owns the Points?

The tracks on this page look like clean coordinates. Each one is the result of choices made by people.

- **Ownership and permission.** The eagle tracks were collected by scientists and shared through Movebank with a license and a citation attached. The Audubon Explorer exists because Audubon asked each data owner for permission. A public data platform depends on people choosing to share.
- **Resolution.** The data in this lab is thinned. A different choice about how many fixes to keep would change the story. How often a tag records, and how many fixes are shared, are design decisions.
- **Risk.** Exact locations of animals can put them at risk, especially for threatened species or at nests and roosts. Who sees a point, and at what level of detail, is a decision for the person building the map.
- **Whose knowledge.** A migration platform shows what instruments recorded. It leaves out what local communities know about the same animals.

These questions connect to the GeoEthics and data sovereignty threads from earlier in the course.

**Think about it:** if the eagle data were from a threatened species, what would you change in your app? Would you keep exact points, show a coarser area, delay the dates, or not publish it? Who should decide?

### Field Notes for Troubleshooting

- **The timeline does nothing.** Go back to the three questions. Check that the field is a date type, that time is enabled on the layer, and that the widgets are connected to the same time-enabled data source.
- **The chart does not change when the timeline moves.** This may be how the widget is connected, or how it works in your version. Test it and record what you see.
- **The MapLibre page is blank.** Open the Console (`F12`). Most problems appear there as red errors. Use Live Preview in VS Code, and do not double-click the HTML file.
- **The Console says `Invalid time value`.** The `timeField` or `idField` in `CONFIG` does not match the property name in `tracks.geojson`. Property names are case-sensitive.
- **The Console says `Chart is not defined`.** A script tag was deleted or mistyped at the bottom of the page.
- **The chart did not change after you pasted.** Make sure you pasted over the old `drawChart` function and did not add a second one, and that no braces were left over.
- **Everything works locally, but the live page is blank.** Check the Pages settings on GitHub and the Console on the live URL.
- **Test your sharing** in a private browser window or on a second device.
