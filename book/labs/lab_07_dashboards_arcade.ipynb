### Part B: Build the required dashboard

Give each element a clear name as you add it. You will pick elements by name when you link them.

| # | Element | Requirement |
|:-:|---|---|
| 1 | **Header** | A title that tells the audience what they are looking at |
| 2 | **Map** | Draws from your web map |
| 3 | **List** | Shows at least two fields for each row, sorted in a way that makes sense for your audience |
| 4 | **Serial chart** | A column or bar chart with a category, a numeric series, and axis titles |
| 5 | **Indicator** | A summary statistic of your choice with a clear title |
| 6 | **Category selector** | Built on your category field and set to filter at least the map, the list, and the indicator |
| 7 | **Two more actions** | For example: the chart pans and zooms the map, the map filters the indicator, or choosing a list row zooms the map |
| 8 | **Theme and access** | Choose a theme, fix any contrast warning, and set the accessible name for every element |

**Test as you go.** Choose a category in the selector: the map, the list, and the indicator should change together. Select a bar in the chart: the map should move.

### Part C: Add Arcade

Choose a tier. **Tier 1 is required.** Tier 2 or Tier 3 is how you reach the top band of the rubric. Whichever tier you choose, give each expression a title, add a comment line that states what it decides, and keep a copy of the text for your submission.

| Tier | What you write | What it adds |
|---|---|---|
| **1 (required)** | A formatting expression on the indicator, list, or table | Color, text, or an icon that responds to the values |
| **2** | A data expression that uses `GroupBy` | A summary by group that feeds a chart or list |
| **3** | A data expression with a loop | A new field the data never had, such as a size or severity class |

**What Tier 1 must include:** a condition (`When`, `IIf`, or `Decode`) tied to a threshold or category you can justify to your audience, and text that carries the same message as the color. Do not rely on color alone.

**A note on field names.** The starter expressions below use placeholder names. Open the Profile Variables panel (Tier 1) or the layer's table (Tiers 2 and 3) and use your layer's real names. Run after every change.

<details>
<summary><b>Tier 1 starter: indicator. Click to expand</b></summary>

```
// Indicator: how busy is it right now?
var n = $datapoint.count_IncidentName;   // use the exact name from the Profile Variables panel
var color = When(n >= 50, '#b8312f', n >= 20, '#d98a00', '#2e7d32');
var label = When(n >= 50, 'High activity', n >= 20, 'Moderate activity', 'Low activity');

return {
  topText: 'Current wildfires in view',
  middleText: Text(n, '#,###'),
  middleTextColor: color,
  bottomText: label,
  bottomTextColor: color
};
```

For Option B, count earthquakes and base the thresholds on a number your audience would care about. For Option C, choose thresholds that make sense for your own data.

</details>

<details>
<summary><b>Tier 1 starter: list. Click to expand</b></summary>

```
// List: shade the biggest rows and add a size class to the template
var size = $datapoint.CalculatedAcres;   // swap in a numeric field from your layer
var sizeClass = When(size >= 10000, 'Major', size >= 1000, 'Large', 'Small');
var shade = When(size >= 10000, '#F3DED7', size >= 1000, '#FBEFD5', '');

return {
  backgroundColor: shade,
  attributes: {
    sizeClass: sizeClass
  }
};
```

Then set the list template to use the new attribute, for example `{IncidentName} ({expression/sizeClass})`.

</details>

<details>
<summary><b>Tier 2 starter: summarize by group. Click to expand</b></summary>

```
// Data expression: incidents and total acres for each state
var fs = FeatureSetByPortalItem(
  Portal('https://www.arcgis.com'),
  'YOUR_ITEM_ID',     // from the item page address, after id=
  0,                  // sublayer number, the number at the end of the layer address
  ['*'],
  false
);

return GroupBy(
  fs,
  ['POOState'],       // the field to group by
  [
    { name: 'incidents', expression: '1', statistic: 'COUNT' },
    { name: 'total_acres', expression: 'IncidentSize', statistic: 'SUM' }
  ]
);
```

To use it, open a chart's data source window, choose **New data expression**, paste the expression, click Run, and name it. Set the chart's category to your group field and its series to one of the new statistics.

</details>

<details>
<summary><b>Tier 3 starter: derive a new field. Click to expand</b></summary>

```
// Data expression: add a size class the layer does not have
var fs = FeatureSetByPortalItem(
  Portal('https://www.arcgis.com'), 'YOUR_ITEM_ID', 0, ['*'], false
);

var rows = [];
for (var f in fs) {
  var acres = IIf(IsEmpty(f.IncidentSize), 0, f.IncidentSize);
  var sizeClass = When(
    acres >= 10000, 'Major (10,000 acres or more)',
    acres >= 1000, 'Large (1,000 to 9,999)',
    'Small (under 1,000)'
  );
  Push(rows, { attributes: { name: f.IncidentName, acres: acres, sizeClass: sizeClass } });
}

var out = {
  fields: [
    { name: 'name', type: 'esriFieldTypeString' },
    { name: 'acres', type: 'esriFieldTypeDouble' },
    { name: 'sizeClass', type: 'esriFieldTypeString' }
  ],
  geometryType: '',
  features: rows
};

return FeatureSet(Text(out));
```

A FeatureSet built this way has no map geometry, so it can feed a chart, list, table, indicator, or selector. The map keeps its own layers. Say what your thresholds are somewhere your audience can see them, for example in a chart title or an element caption.

</details>

Keep these prompts in mind as you work. Jot notes if that helps, and nothing here is collected. They feed straight into your reflection questions.

- **Advanced formatting:** When you switched it on, what happened to the settings that were already there?
- **Errors:** What did the editor tell you when something was wrong, and what did you check first?

- **Actions:** Which link between elements surprised you? How would you have wired the same behavior in MapLibre?

**Pacing guide** (adjust to your schedule). The core build with Tier 1 takes about 2 hours and 30 minutes. Tier 2 or Tier 3 adds roughly 45 minutes.

| 2 (about 60 min) | Elements 5 to 8 | Indicator and selector working, three actions tested, theme and accessible names set |

**Working rules:**
- Save as you go, and run your expression after every small change.
- If something does not behave, treat it as a troubleshooting moment. Read the message, check the Profile Variables panel, and look the function up in the reference before you ask. Keep a note of what you tried, because you will describe one of these moments in your reflection.
- Post questions on the discussion board so the answers help everyone.
