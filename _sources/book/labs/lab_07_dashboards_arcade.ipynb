### Part C: Add Arcade

You need **one working Arcade expression** that formats the indicator, list, or table. Give it a title, add a comment line that states what it decides, and keep a copy of the text for your submission.

**What your expression must include:** a condition (`When`, `IIf`, or `Decode`) tied to a threshold or category you can justify to your audience, and text that carries the same message as the color. Do not rely on color alone.

**Guided track (Option A).** The tutorial already gives you expressions for the list and the indicator. They are your starting point. Make one change of your own to one of them, and mark it with a comment such as `// CHANGED: ...`. Ideas: move a threshold and explain why, add a third severity class with a text label, add a field such as the cause to the list template, or check how the expression behaves for planned outages with a start time in the future and fix it if it looks wrong. Finishing the tutorial without a change of your own does not meet the Arcade criterion.

**Open track (Options B, C, D).** Adapt one of the starters below to your layer.

**A note on field names.** The starter expressions use placeholder names. Open the Profile Variables panel and use your layer's real names. Run after every change.

<details>
<summary><b>Starter: indicator. Click to expand</b></summary>

```
// Indicator: how busy is it right now?
var n = $datapoint.count_IncidentName;   // use the exact name from the Profile Variables panel
var color = When(n >= 50, '#b8312f', n >= 20, '#d98a00', '#2e7d32');
var label = When(n >= 50, 'High activity', n >= 20, 'Moderate activity', 'Low activity');

return {
  topText: 'Current wildfires in view',
  middleText: Text(n, '#,###'),
  middleTextColor: color,
  bottomText: label,
  bottomTextColor: color
};
```

This starter uses wildfire field names (Option B). For Option C, count earthquakes and base the thresholds on a number your audience would care about. For Option D, choose thresholds that make sense for your own data.

</details>

<details>
<summary><b>Starter: list. Click to expand</b></summary>

```
// List: shade the biggest rows and add a size class to the template
var size = $datapoint.CalculatedAcres;   // swap in a numeric field from your layer
var sizeClass = When(size >= 10000, 'Major', size >= 1000, 'Large', 'Small');
var shade = When(size >= 10000, '#F3DED7', size >= 1000, '#FBEFD5', '');

return {
  backgroundColor: shade,
  attributes: {
    sizeClass: sizeClass
  }
};
```

Then set the list template to use the new attribute, for example `{IncidentName} ({expression/sizeClass})`.

</details>

Permitted uses can include using AI to help troubleshoot a step that is not working in ArcGIS Dashboards, to explain an Arcade error message in plainer language, or to explain a term or function from Esri's documentation. AI tools often suggest Arcade functions or dictionary keys that do not exist, so anything an AI suggests must be checked against Esri's reference or by running it in the editor, and you should be ready to describe how you checked it. However, independent installation and use of AI agents to complete GIS tasks is not permitted at this time. AI may also be used to support general writing procedures (outlining a response, improving clarity or grammar, etc.), but your reflection must remain original to you as the sole responsible author, and it must describe your own experience building your own dashboard.
