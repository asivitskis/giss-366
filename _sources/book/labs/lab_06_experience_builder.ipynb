### Part B: Complete the Esri exercises, in order

The seven exercises build on each other, and you will keep working on **one single experience** from start to finish. Do them in order, and do not delete or restart your experience between sittings. The times below are the estimates Esri lists in the course; give yourself extra time if you are new to Experience Builder.

| # | Esri exercise | About | Workflow step it mainly practices |
|:-:|---|:-:|---|
| 1 | Use ArcGIS Experience Builder to begin creating a web app | 20 min | Plan, Design, start of Configure |
| 2 | Configure page element widgets in ArcGIS Experience Builder | 30 min | Configure |
| 3 | Configure mapcentric widgets in ArcGIS Experience Builder | 30 min | Configure, Customize |
| 4 | Add an informational window to an experience | 20 min | Customize |
| 5 | Apply a theme to a web app using ArcGIS Experience Builder | 20 min | Customize |
| 6 | Optimize a web app using ArcGIS Experience Builder | 30 min | Customize, Iterate |
| 7 | The final exercise: deploying and testing the experience | 15 min | Deploy, Iterate |

**Total:** roughly 2 hours and 45 minutes.

**Working rules:**
- Follow Esri's directions and naming conventions for your experience exactly. I will use the name to find your item in our organization.
- Save as you go. If you leave and come back, reopen the same experience from the Experience Builder home page.
- Use the course's built-in "view result" and "show answer" prompts to check yourself. They are for you; I do not collect them.
- If an exercise does not behave the way the instructions say, treat it as a troubleshooting moment: re-read the previous step, check the documentation, and then ask. Keep a note of what you tried, because you will describe one of these moments in your reflection.

## Step 3: Act

### What to turn in (graded: Lab Exercises)

1. The **link to your published experience** (the final exercise shows you how to get it).
2. Confirmation that your experience is **shared with the organization**, so that I can open it and see your work. If you are unsure how to check, use the sharing guidance in the final exercise, and tell me in your submission if something stopped you.
3. A short written reflection (~400 words total, roughly 60 to 80 words per question) answering:
   - **Prediction check:** Revisit your Step 1 predictions. How close was your time estimate, and which features did you correctly predict the builder would handle for you?
   - **Hand-built vs. builder-built:** Pick one behavior in your finished app (for example, something that updates when a user selects something else). Describe how you would have built the same behavior in Leaflet or MapLibre, and name one thing the builder made easier and one thing it decided for you.
   - **Workflow:** Which step of the Experience Builder workflow did you spend the most time on, and which did you underestimate? Use the step names from Tuesday's session.
   - **Widgets and actions:** Using one widget you configured as your example, explain the difference between its content, style, and action settings. Say whether one behavior in your app was a message action or a data action, and how you could tell.
   - **Reach and access:** Who can open your published experience right now? What has to be true about the web map and data behind it for that to work? Name one accessibility or device decision you made or saw in the exercises, how you tested it, and one barrier a real audience could still face.
   - **Troubleshooting and the continuum:** Describe one specific moment you got stuck and how you got unstuck. Then place Experience Builder on the continuum from Tuesday (no-code builder, low-code customization, developer edition, Maps SDK, open-source libraries) and say what you would have to learn or pay for to move one step toward more control.

You do **not** need to submit screenshots of individual steps. I can see what you completed in your ArcGIS Online account, and I will check the link you submit.
