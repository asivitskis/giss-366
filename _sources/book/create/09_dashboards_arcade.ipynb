Last week you assembled an experience from widgets. A dashboard uses the same architecture underneath: data sits on a server, and a browser client asks for it and draws it. What changes is the job the screen does. An experience invites someone to explore or follow a story. A dashboard puts several views of the same data on one screen so someone can monitor what is happening, make a decision, or spot a trend at a glance.

## What Is a Dashboard For?

A dashboard is a presentation of geographic information and data that helps you monitor events, make decisions, inform others, and see trends. Its elements work together on a single screen. Esri groups dashboards by purpose, and the right purpose follows from who is looking and what they need to do next.

| Type | Question it answers | Example |
|---|---|---|
| Operational | What is happening right now? | Current wildfire incidents on a map, with a count |
| Strategic | How are we doing against our goals? | A county office tracking service targets |
| Analytical | What trends and patterns are in the data? | Fire counts by year, compared with rainfall |
| Informative | What story does the data tell? | A public page explaining a fuels treatment project |

An effective dashboard does four things:

1. It draws attention to where it is needed.
2. It shows what matters most on a screen full of data.
3. It lets its audience understand what is happening and respond quickly.
4. It states measures clearly, accurately, and without distraction.

Dashboards also differ by how they are viewed. Some run unattended on a large screen in an operations center and offer a passive experience. Others are attended on a desktop monitor and invite more interaction.

## The Parts of a Dashboard

**Elements** are the building blocks. Common ones are maps, lists, charts, gauges, indicators, and tables. Most elements are data driven, which means each one needs a data source: a layer, a web map's layers, or a data expression (more on those below).

A practical habit: add a **map element first**. A web map's operational layers then show up as data sources when you configure the other elements.

Every element is configured through a few areas: a header for a title and more information, an optional top caption and bottom caption, the visualization itself, and an accessibility tab where you set the name a screen reader will read. Give each element a clear name early, because those names are what you pick from when you link elements together.

**Layout.** A view is built from rows and columns, and it always fills the screen. You can dock, resize, stack (tabs), and group elements. A view can also have one header and one sidebar.

**Views.** The desktop view is created first. A mobile view is a second layout for small screens, with its own arrangement of the same or different elements. Phones load the mobile view automatically when the screen is narrow, and a dashboard without one simply uses the desktop view. Phones have less room and memory, so a mobile view usually holds fewer elements.

## Actions: How Elements Talk to Each Other

One feature turns a set of charts on a page into a dashboard: **actions**. When actions are enabled, interacting with one element updates another. In Esri's wildfire tutorial, two links do most of the work:

- The **serial chart** is set to pan and zoom the map, so choosing a bar takes you to that fire.
- The **map** is set to filter the indicator, so the count shows only the fires inside the current map extent.

We practiced building these two links in class. If you have written an event listener in JavaScript, an action is the same idea without writing the code. It also matches the message actions you configured in Experience Builder last week.

## Design and Access

A **theme** is a set of colors and styling options saved with the dashboard and applied to both views. The light theme is the default, and you can choose a provided theme or customize one. A few basic colors drive most of the design, and the others are calculated from them. When the dashboard has a map, matching a light or dark basemap to the theme keeps the screen consistent.

Three habits keep a dashboard readable for the widest range of people:

- **Check contrast.** The color picker shows a warning when a text and background pairing fails a contrast check. Fix it instead of ignoring it.
- **Set accessible names.** Each element has an accessibility tab. The name you enter is what a screen reader announces, and it should be close to the element's title.
- **Use the glance test.** Look at your dashboard for five seconds, then look away. What do you remember? If it is the wrong thing, rearrange the screen so that the most important element is the most prominent one.

## Copying a Dashboard from Its Item ID

Every dashboard is an item with an ID. That makes it possible to start a new dashboard from an existing one, and it is a quick way to learn how a dashboard was built.

The pattern, as described in Esri's first tutorial:

1. Start creating a new dashboard (for example, from the **Create app** menu on a saved web map, then choose **Dashboards**).
2. In the address of the new dashboard, replace the text after `#id=` with the item ID of the dashboard you want to start from, and press Enter.
3. A **data source** window opens, with the original map, layers, and fields in one column and your replacements in the other.
4. Choose your own map and layers. Some fields have no one-to-one match in your data. For those, pick a different metric to show or delete the element.
5. If an element shows a ratio, set both a **value field** (the top of the fraction) and a **reference field** (the bottom).

This works when the creator has allowed it, so check the sharing settings, and a copy only shows data that you are permitted to see. A few questions to ask whenever you build on someone else's dashboard:

- Did I credit the person or organization whose work I started from?
- Am I allowed to reuse the data behind it?
- Which of their design choices did I inherit, and do they fit my audience?

## A First Look at Arcade

**Arcade** is a portable, lightweight, and secure expression language. It can do math, work with text, and evaluate logic. You may have seen it already in web maps, where it writes custom pop-ups, labels, and styles. Dashboards add two more uses.

| | Data expressions | Formatting expressions |
|---|---|---|
| What it does | Builds a new data source for the dashboard (combine, model, or enhance existing data) | Controls how an element looks, based on its data |
| What it returns | A **FeatureSet** | A **dictionary** of key-value pairs |
| Used by | Data-driven elements | Indicator, list, and table elements |
| Runs | When the data loads | Once for each data point the element draws |

Formatting is the easiest place to start. Say an indicator shows the number of current wildfires, and you want the number to turn red when the count is high. A formatting expression looks at the value, makes a decision, and returns the colors and text to display. Esri's advanced formatting page lists the property names an indicator accepts, and the expression below uses real ones:

```
// Arcade formatting expression for an indicator
var n = $datapoint.count_IncidentName;   // confirm this name in the Profile Variables panel
var color = When(n >= 50, '#b8312f', n >= 20, '#d98a00', '#2e7d32');
var label = When(n >= 50, 'High activity', n >= 20, 'Moderate activity', 'Low activity');

return {
  topText: 'Current wildfires in view',
  middleText: Text(n, '#,###'),
  middleTextColor: color,
  bottomText: label,
  bottomTextColor: color
};
```

Three details to notice:

1. The expression runs once per data point. An indicator draws one data point. A list draws one per row.
2. What `$datapoint` contains depends on how the element is set up. An indicator built on a statistic receives a dictionary whose keys are the statistic names (Esri's page shows `avg_Temperature` for the average of a Temperature field). An element built on features receives a feature with that layer's fields. The Profile Variables panel in the editor shows what you really have, so look there before you type a name.
3. The color travels with a text label, so the message does not depend on color alone.

## Arcade and the JavaScript You Know

As we have written JavaScript in this course, some of Arcade will feel familiar. Treat the match as a helpful similarity, with real differences.

| Looks familiar | Works differently |
|---|---|
| Variables, with `var` | Arcade runs inside ArcGIS applications, so there is no web page or DOM to change |
| `if` and `else` statements | There are no event listeners. The application decides when your expression runs |
| Functions and `return` | What you return has to match the structure the element expects (a dictionary, or a FeatureSet) |
| Comments with `//` | Many built-in functions are specific to GIS, such as ones that read feature attributes and filter layers |

When you meet a function you do not know, look it up in the Arcade function reference. Reading the reference is part of the skill, and it is how working GIS staff learn new functions too.

**Live in this notebook:** the cell below uses Python to show the same decision logic that the indicator expression above uses. The function reads a count, makes a decision, and returns a dictionary of display settings. The key names match the ones Esri's indicator formatting accepts, so the shape of the result is the same one Arcade returns.

````python
from IPython.display import HTML, display

def indicator_format(n):
    """Return a dictionary of display settings, the way an Arcade formatting expression does."""
    if n >= 50:
        color, label = "#b8312f", "High activity"
    elif n >= 20:
        color, label = "#d98a00", "Moderate activity"
    else:
        color, label = "#2e7d32", "Low activity"
    return {
        "topText": "Current wildfires in view",
        "middleText": f"{n:,}",
        "middleTextColor": color,
        "bottomText": label,
        "bottomTextColor": color,
    }

def show_indicator(n):
    f = indicator_format(n)
    display(HTML(
        f"<div style='font-family: Georgia, serif; border: 1px solid #999; padding: 10px; width: 220px; background: #f4f8fb;'>"
        f"<div style='font-size: 13px; color: #333;'>{f['topText']}</div>"
        f"<div style='font-size: 40px; font-weight: bold; color: {f['middleTextColor']};'>{f['middleText']}</div>"
        f"<div style='font-size: 13px; color: {f['bottomTextColor']};'>{f['bottomText']}</div></div>"
    ))

for count in (8, 27, 63):
    show_indicator(count)

# Try it: change the thresholds, or add a fourth level with its own color and label.
````

## Anatomy of an Arcade Expression

**Inputs, expression, result.** An Arcade expression works like a formula in a spreadsheet cell. You give it inputs (values from your data), it evaluates them, and it hands back a result. Every expression has to return a value. What that value must look like depends on where the expression runs: a dictionary for a formatting expression, a FeatureSet for a data expression.

**Profiles.** Arcade runs in many places, including pop-ups, labels, field calculations, forms, and dashboards. Each place is a *profile*. The profile decides which variables you can read (`$feature` in pop-ups and labels, `$datapoint` in dashboard formatting), which function bundles exist, and what the expression has to return. Two questions come before the first line of any expression:

1. What can I read here?
2. What must I return here?

**Three language features** make up most expressions:

- **Functions** are preset actions. They cover text, math, dates, arrays, and features. A core bundle is available in every profile, and some profiles add bundles for data access, geometry, or portal items.
- **Variables** hold values under a name. You create one with `var`. Profile variables start with `$` and give you access to the data. Constants such as pi are built in.
- **Loops** repeat a block of code. A `for` loop visits each item in an array or each feature in a FeatureSet. A `while` loop repeats as long as a condition holds.

**The editor** gives you three tools you will use constantly. The **Run** button evaluates your expression with sample data and shows the output. The **Console** function writes a value of your choosing to a console tab, which lets you see what is happening in the middle of an expression. The **side panel** lists the profile variables and functions that exist in the current profile, with a short help entry for each function. The [Arcade Playground](https://developers.arcgis.com/arcade/playground/) is a free web page for testing logic without opening a dashboard.

## Reading an Expression One Line at a Time

Here is the indicator expression from earlier, taken apart:

| Line | What happens | Language feature |
|---|---|---|
| `var n = $datapoint.count_IncidentName;` | Reads the statistic from the data point and stores it under the name `n` | Profile variable, variable |
| `var color = When(n >= 50, '#b8312f', ... , '#2e7d32');` | Tests the conditions in order and returns the value for the first one that is true. The last value is the default | Function |
| `var label = When(...);` | Does the same for the text label | Function |
| `return { ... };` | Builds a dictionary of display properties and hands it back to the indicator | Return, dictionary |
| `middleText: Text(n, '#,###')` | Converts the number to text and adds thousands separators | Function, number pattern |

| You need to | Function | Example |

## Formatting Expressions: What Each Element Accepts

Esri's [advanced formatting page](https://doc.arcgis.com/en/dashboards/latest/create-and-share/advanced-formatting.htm) is the reference for the full property lists. The main ideas are below.

**Indicator.** The expression returns a dictionary. The text sits in three bands (top, middle, bottom), and each band has its own text and color properties, with extra properties for outline, weight, and size. There are also properties for the overall text color and background color, an icon (you add the icon in the area below the editor, give it a name, and return that name as `iconName`), icon color and placement, and a `noValue` flag that shows "no value" text in place of the indicator. When you switch on advanced formatting, the element's normal color and value-format settings turn off, so the expression has to do that work.

**List.** The expression runs once for each line item. It returns colors for the text, background, separator line, and selected state.

**Table.** The expression returns a `cells` dictionary with one entry for each column. Each cell needs `displayText` and can also set its own colors, alignment, and icon.

**Your own attributes.** Any expression can return an `attributes` dictionary. Whatever you put there can be reused in the element's template, title, or captions by writing `{expression/yourName}`. This is how a list can show a computed size class next to each incident name.

**Details that cause errors:**

- Colors must be strings: hex (`'#b8312f'`), `rgb(...)`, `rgba(...)`, `hsl(...)`, or `hsla(...)`.
- Put a comma after every property in the dictionary except the last one.
- Field names with spaces or special characters cannot be read, so use the name the Profile Variables panel gives you.
- Geometry functions are not available in list formatting.
- If your expression uses `Now()`, turn on the **Refresh script** setting and choose an interval so the result stays current.

## Data Expressions: Building a New Data Source

A data expression builds a new data source for an element. It always returns a **FeatureSet**, and elements can choose it in the data source window the same way they choose a layer. The usual pattern has two steps: load a layer by its item ID with `FeatureSetByPortalItem`, then transform it.

`FeatureSetByPortalItem` takes the portal address, the item ID, the sublayer number, a list of fields (`['*']` means all), and whether to include geometry. The item ID is in the web address of the item page, after `id=`. The sublayer number is the number at the end of the layer's service address.

**Example 1: summarize by group.** This expression counts incidents and totals acres for each state. The result has one row per state, which makes it ready for a serial chart.

```
var fs = FeatureSetByPortalItem(
  Portal('https://www.arcgis.com'),
  'YOUR_ITEM_ID',     // from the item page address, after id=
  0,                  // sublayer number
  ['*'],
  false
);

return GroupBy(
  fs,
  ['POOState'],       // the field to group by; confirm the name in your layer
  [
    { name: 'incidents', expression: '1', statistic: 'COUNT' },
    { name: 'total_acres', expression: 'IncidentSize', statistic: 'SUM' }
  ]
);
```

**Example 2: derive a field the data never had.** The layer has acres but no size class. A loop can build one. The loop reads each feature, decides its class, and collects the rows. The last line turns the collected rows into a FeatureSet.

```
var fs = FeatureSetByPortalItem(
  Portal('https://www.arcgis.com'), 'YOUR_ITEM_ID', 0, ['*'], false
);

var rows = [];
for (var f in fs) {
  var acres = IIf(IsEmpty(f.IncidentSize), 0, f.IncidentSize);
  var sizeClass = When(
    acres >= 10000, 'Major (10,000 acres or more)',
    acres >= 1000, 'Large (1,000 to 9,999)',
    'Small (under 1,000)'
  );
  Push(rows, { attributes: { name: f.IncidentName, acres: acres, sizeClass: sizeClass } });
}

var out = {
  fields: [
    { name: 'name', type: 'esriFieldTypeString' },
    { name: 'acres', type: 'esriFieldTypeDouble' },
    { name: 'sizeClass', type: 'esriFieldTypeString' }
  ],
  geometryType: '',
  features: rows
};

return FeatureSet(Text(out));
```

**Things to know before you use one:**

- A FeatureSet built this way has no map geometry, so it can feed a chart, list, table, indicator, or selector. The map keeps using the web map's own layers.
- Derived values exist only inside the dashboard. The thresholds are your decisions, so say what they are somewhere your audience can see.
- Both examples need a real item ID and real field names. The editor's Run button shows the output table, so check it before you attach the expression to an element.
- Esri's support article on [calculating a mean and using GroupBy in a dashboard table](https://support.esri.com/en-us/knowledge-base/how-to-calculate-the-mean-and-use-the-groupby-function--000027670) shows the same pattern with a table element.

````python
from collections import defaultdict

# A tiny made-up table that stands in for a layer (these are not real incidents)
incidents = [
    {"name": "Example Fire A", "state": "NM", "acres": 12400},
    {"name": "Example Fire B", "state": "NM", "acres": 850},
    {"name": "Example Fire C", "state": "AZ", "acres": 3100},
    {"name": "Example Fire D", "state": "AZ", "acres": 40},
    {"name": "Example Fire E", "state": "CO", "acres": 15200},
]

# Data expression idea 1: derive a field the data never had (like the loop in Example 2)
def size_class(acres):
    if acres >= 10000:
        return "Major"
    if acres >= 1000:
        return "Large"
    return "Small"

for row in incidents:
    row["sizeClass"] = size_class(row["acres"])

# Data expression idea 2: summarize by group (like GroupBy in Example 1)
summary = defaultdict(lambda: {"incidents": 0, "total_acres": 0})
for row in incidents:
    summary[row["state"]]["incidents"] += 1
    summary[row["state"]]["total_acres"] += row["acres"]

print("Derived field:")
for row in incidents:
    print(f"  {row['name']:<16}{row['acres']:>8,}  {row['sizeClass']}")

print("\nGrouped by state:")
for state, vals in summary.items():
    print(f"  {state}  incidents={vals['incidents']}  total_acres={vals['total_acres']:,}")

# Try it: change a threshold and rerun. Which fire changes class? Who should get to decide where 'Major' starts?
````

## Working Habits for Writing Arcade

1. **Run early and often.** Run after every small change. A short expression that works is a better base than a long one that fails.
2. **Read the error before you change anything.** The message usually names a line and a problem. Fix that one thing and run again.
3. **Look at what you have.** Open the Profile Variables panel to see the exact field and statistic names.
4. **Log the middle.** `Console(n)` shows what a variable holds at that point. When the result looks wrong, log the values that feed it.
5. **Look it up.** Check each unfamiliar function in the reference or the editor's help entry.
6. **Name and comment.** Give each expression a title and a one-line comment that states its thresholds, so that you and your audience can see what it decides.
7. **Verify outside help.** If a person or a tool hands you an expression, run it and check every function and key against the reference before you rely on it.

```
  topText: 'Current wildfires in view'
  middleText: Text(n, '#,###'),
  middleTextColor: colour,
  bottomText: 'Fires in view'
};
```

When it works, add `Console(n);` on a new line after the first one, run it, and find the **Console** tab.

*Record:* Which errors did you find, and what in the editor's message pointed you to each one?

### Card 4: List it (early finishers, about 5 minutes)

Open your list element, go to the **List** tab, and enable advanced formatting. Paste this expression, then swap in your acres field name from the Profile Variables panel.

```
var acres = $datapoint.CalculatedAcres;   // use the acres field your chart uses
var sizeClass = When(acres >= 10000, 'Major', acres >= 1000, 'Large', 'Small');
var shade = When(acres >= 10000, '#F3DED7', acres >= 1000, '#FBEFD5', '');

return {
  backgroundColor: shade,
  attributes: {
    sizeClass: sizeClass
  }
};
```

Then set the list's template to show the incident name and the new attribute, for example `{IncidentName} ({expression/sizeClass})`.

*Record:* Which rows are shaded? Do the thresholds seem right for this data, and who should decide?

## Looking Ahead: Lab 7

Lab 7 is a semi-open dashboard build. You choose a data option (current wildfires, recent earthquakes, or your own layer), build a dashboard with required elements that talk to each other, and add Arcade at the level you choose. Thursday's session introduces the lab and works through Arcade together.

| Tier | What you write | What it adds to the dashboard |
|---|---|---|
| **1 (required)** | A formatting expression on an indicator, list, or table | Color, text, or an icon that responds to the values |
| **2** | A data expression that uses `GroupBy` | A summary by group, such as acres by state, that feeds a chart or list |
| **3** | A data expression with a loop | A new field the data never had, such as a size or severity class |

A working Tier 2 or Tier 3 expression puts your Arcade work in the top band of the rubric. The full instructions are on the Lab 7 page.
